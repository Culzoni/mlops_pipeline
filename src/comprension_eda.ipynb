```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargamos los datos
df = pd.read_csv("../Base_de_datos.csv")

# 2. Resumen de las columnas
print("--- Información General ---")
df.info()

# 3. Conteo por Plan (en minúscula)
print("\n--- Clientes por Plan ---")
print(df['plan'].value_counts())

# 4. Gráfico de Churn (en minúscula)
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='churn')
plt.title('Distribución de Abandono de Clientes (0 = Se quedan, 1 = Cancelan)')
plt.show()
```
