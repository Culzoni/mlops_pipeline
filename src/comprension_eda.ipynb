Analisis Univariable 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargamos los datos
df = pd.read_csv("../Base_de_datos.csv")

# 2. Resumen de las columnas
print("--- Información General ---")
df.info()

# 3. Conteo por Plan (en minúscula)
print("\n--- Clientes por Plan ---")
print(df['plan'].value_counts())

# 4. Gráfico de Churn (en minúscula)
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='churn', hue='churn', palette=['blue', 'orange'], legend=False)
plt.title('Distribución de Abandono de Clientes (0 = Se quedan, 1 = Cancelan)')
plt.show()
```

Analisis Bivarible 

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,4))
sns.boxplot(data=df, hue='churn', y='age', palette=['green', 'red'], legend=False)
plt.title('Analisis Bivariable: Edad vs Abandono de Clientes')
plt.xlabel('churn (0 = Siguen, 1 = Cancelan)')
plt.ylabel('Edad')
plt.show()
```

Analisis Multivariable 

```python

plt.figure(figsize=(10, 6))

columnas_num = df.select_dtypes(include=['int64', 'float64']).columns
sns.heatmap(df[columnas_num].corr(), annot=True, cmap='coolwarm', fmt=".2f")

plt.title('Análisis Multivariable: Mapa de Calor de Correlaciones')
plt.show()
```
